## 1. From optimal control to Hamiltonian dynamics

For $\dot{x}=f(x,u,t)$, Pontryagin's minimum principle introduces

$$H(x,p,u,t)=L(x,u,t)+p^\top f(x,u,t).$$

After pointwise minimization over $u$, the state and costate satisfy

$$\dot{x}=\partial_p H, \qquad \dot{p}=-\partial_x H.$$

Unknown initial costates, switching instants, and free terminal times are recovered with shooting methods.

## 2. Methods compared

- **HPI:** an implicit, groupoid-derived second-order Hamiltonian Poisson step.
- **RK2:** an explicit second-order midpoint Runge-Kutta step.

Both propagators are used inside nonlinear shooting procedures. Outer-solver evaluation counts are not total computational costs because each HPI step includes an internal fixed-point iteration.

## 6. Limitations and next step

These five systems use canonical state-costate spaces. They test the symplectic limit of the construction rather than uniquely Poisson features such as preservation of nonconstant Casimirs or nontrivial symplectic leaves.

A stronger follow-up should use a genuinely noncanonical benchmark and compare endpoint error, objective error, runtime, Hamiltonian drift, Casimir drift, and leaf error under mesh refinement.
